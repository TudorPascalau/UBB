## Exercitiul 1

```python
from scipy.stats import bernoulli, binom
from matplotlib.pyplot import bar, show, hist, grid, legend, xticks

def mers(n, p):
    poz = 0
    positions = []
    for i in range(n):
        if bernoulli.rvs(p) == 1:
            poz += 1
        else:
            poz -= 1
            
        positions.append(poz)
    
    return poz

print(mers(10, 0.5))
```

```python
N = 1000
n = 10
p = 0.5

data = [mers(n, p) for _ in range(N)]
bin_edges = [k+0.5 for k in range(-n-1, n+1)]
hist(data, bin_edges, density = True, rwidth = 0.9, color = 'green', edgecolor = 'black',alpha = 0.5, label = 'frecvente relative')

distrib = dict([(2*i-n, binom.pmf(i, n, p)) for i in range(0, n+1)])
bar(distrib.keys(), distrib.values(), width = 0.75, color = 'red', edgecolor = 'black', alpha = 0.4, label = 'valori teoretice')
xticks(range(-n-1, n+1))
legend(loc = 'upper right')
grid()
show()

# X = S1 + ... + Sn
# X = (2B1 - 1) + ... + (2Bn - 1)
# X = 2B - n
# B ~ Bino(n, p)
```

```python
def mers_cerc(n, noduri, p):
    poz = 0
    positions = []
    for i in range(n):
        if bernoulli.rvs(p) == 1:
            poz += 1
        else:
            poz -= 1

        if(poz == -1):
            poz = noduri - 1

        if(poz == noduri):
            poz = 0

        positions.append(poz)
    
    return positions

print(mers_cerc(10, 5, 0.5))
```
