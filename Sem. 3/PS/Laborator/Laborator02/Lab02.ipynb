1. Doi jucatori aleg doua sume distincte care se pot obttine prin aruncarea a trei zaruri, apoi simuleaza 
în Python aruncarea a trei zaruri de 100000 de ori. Jucatorul câstigator este cel a carui suma apare
mai des. Ce sume distincte ar trebui sa aleaga cei doi jucatori pentru a-si maximiza sansele de câstig?

```python

contor = {i: 0 for i in range(3,19)}
for _ in range(1000000):
    z = [randint(1, 6) for _ in range(3)]
    suma = sum(z)
    contor[suma] += 1

print(contor)
```

```python
plot(contor.keys(), contor.values(), '*')
xticks(range(3, 19))
grid()
show()
```

```python
contor = {i: 0 for i in range(3,19)}
for z in product(range(1, 7), repeat=3):
    suma = sum(z)
    contor[suma] += 1

print(contor)

plot(contor.keys(), contor.values(), '*')
xticks(range(3, 19))
grid()
show()
```

i) sunt în interiorul cercului tangent laturilor patratului.

```python
axis('square')
axis((0, 1, 0, 1))

count = 0
total = 2500

for _ in range(total):
    F=[random(),random()]
    E = [0.5, 0.5]

    if dist(F, E) < 0.5:
        plot(F[0], F[1], 'bo')
        count += 1
    else:
        plot(F[0], F[1], 'ro')
print(count, total)
print(count/total)
print(pi*0.5*0.5 / 1)
print(count/total*4)
show()
```

ii) sunt mai apropiate de centrul patratului decât de vârfurile patratului

```python
axis('square')
axis((0, 1, 0, 1))

count = 0
total = 2500

E = [0.5, 0.5]
A = [0, 0]
B = [0, 1]
C = [1, 1]
D = [1, 0]

for _ in range(total):
    F=[random(),random()]

    if dist(F, E) < dist(F, A) and dist(F, E) < dist(F, B) and dist(F, E) < dist(F, C) and dist(F, E) < dist(F, D):
        plot(F[0], F[1], 'bo')
        count += 1
    else:
        plot(F[0], F[1], 'ro')
show()
print(count)
print(count/total)
```

iii) sunt mai apropiate de centrul patratului decât de latura de jos a patratului.

```python
axis('square')
axis((0, 1, 0, 1))

count = 0
total = 2500

E = [0.5, 0.5]

for _ in range(total):
    F = [random(), random()]

    if dist(F, E) < F[1]:
        plot(F[0], F[1], 'bo')
        count += 1
    else:
        plot(F[0], F[1], 'ro')
show()
print(count)
print(count/total)
```
